```python
# ===================================================================================
# [최종] YOLO + EasyOCR 연동 코드 (중간 파일 저장 필요 없음)
# ===================================================================================

import os
from PIL import Image
from google.colab import drive
import cv2

# --- 1. 환경 설정 및 라이브러리 설치 ---
print(">>> 1단계: 환경 설정 및 라이브러리 설치를 시작합니다...")
# YOLO와 EasyOCR 라이브러리를 설치합니다.
!pip install ultralytics easyocr &> /dev/null
from ultralytics import YOLO
import easyocr

# 구글 드라이브 마운트
if not os.path.exists('/content/drive/MyDrive'):
    drive.mount('/content/drive', force_remount=True)
print("✅ 라이브러리 설치 및 준비 완료!")


# --- 2. 모델 로드 ---
print("\n>>> 2단계: 모델들을 불러옵니다...")
try:
    # 2-1. 학습된 YOLO 모델 로드
    yolo_model_path = '/content/drive/MyDrive/Colab Notebooks/ESG_B 딥러닝/yolo_results/drug_detection_run_final/weights/best.pt'
    yolo_model = YOLO(yolo_model_path)
    print("  - ✅ YOLO 모델 로드 완료!")

    # 2-2. EasyOCR 리더기 초기화
    reader = easyocr.Reader(['ko', 'en'])
    print("  - ✅ EasyOCR 리더기 초기화 완료!")

except Exception as e:
    print(f"🚨 모델 로드 실패: {e}")
    raise SystemExit("모델 파일 경로를 확인하거나 라이브러리 설치를 다시 확인해주세요.")


# --- 3. 테스트 실행 ---
print("\n>>> 3단계: 새로운 이미지로 테스트를 실행합니다...")
# 테스트할 전체 이미지 경로
test_image_path = '/content/drive/MyDrive/Colab Notebooks/ESG_B 딥러닝/data/yolo_data/images/medicine_00001.jpg'

if not os.path.exists(test_image_path):
    print(f"🚨 테스트 이미지를 찾을 수 없습니다: {test_image_path}")
else:
    print(f"✅ 테스트 이미지: {os.path.basename(test_image_path)}")

    # 3-1. YOLO로 라벨 위치 감지
    yolo_results = yolo_model(test_image_path, verbose=False)

    if len(yolo_results[0].boxes) > 0:
        original_image = cv2.imread(test_image_path)

        for i, box in enumerate(yolo_results[0].boxes):
            # 3-2. 라벨 영역을 '데이터'로 잘라내기 (파일 저장 X)
            xyxy = box.xyxy[0].cpu().numpy().astype(int)
            x1, y1, x2, y2 = xyxy
            h, w, _ = original_image.shape
            x1, y1, x2, y2 = max(0, x1), max(0, y1), min(w, x2), min(h, y2)
            cropped_image_data = original_image[y1:y2, x1:x2]

            # 3-3. 잘라낸 '데이터'를 EasyOCR로 전달
            ocr_results = reader.readtext(cropped_image_data)

            if ocr_results:
                extracted_text = " ".join([res[1] for res in ocr_results])
            else:
                extracted_text = "인식된 텍스트가 없습니다."

            print("\n" + "="*50)
            print(f"   감지된 라벨 #{i+1} 최종 결과")
            print("="*50)
            print(f"▶️ 인식된 텍스트: {extracted_text}")
            print("\n▶️ 라벨 이미지:")
            display(Image.fromarray(cv2.cvtColor(cropped_image_data, cv2.COLOR_BGR2RGB)))
            print("="*50)

    else:
        print("해당 이미지에서 라벨을 찾지 못했습니다.")
```
